```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("train.csv")
sex_count = df["Sex"].value_counts()
plt.bar(sex_count.index, sex_count, color=["lightgreen","skyblue"])
plt.title("Bar : Number of Gender")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.grid(True,linestyle="-",alpha=0.7)
```
