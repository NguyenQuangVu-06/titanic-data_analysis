```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv('train.csv')
plt.scatter(df["Age"],df["Fare"], color="lightblue", edgecolor="black",alpha=0.6)
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Scatter of Age and Fare" , color="yellow" , fontsize=15)
plt.grid(True,linestyle="--")
```
