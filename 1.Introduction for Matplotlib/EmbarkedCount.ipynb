```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("train.csv")
embarked = df["Embarked"].dropna().value_counts()
plt.bar(embarked.index,embarked.values,color=["purple","yellow","skyblue"])
plt.set(title="Bar : Count of Embarked" , xlabel="Emabarked" , ylabel="Number")
plt.xticks(["S","C","Q"],["Cảng S","Cảng C","Cảng Q"])
```
