```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("train.csv")
def fare_group(x):
    if x < 100:
        return "Cheap"
    elif x < 250 :
        return "Medium"
    else :
        return "High"
df["FareGroup"] = df["Fare"].apply(fare_group)
fare_group=df["FareGroup"].value_counts()
plt.bar("Cheap" , fare_group["Cheap"] , label="Cheap" , color = "yellow")
plt.bar("Medium" , fare_group["Medium"] , label="Medium" , color = "purple")
plt.bar("High" , fare_group["High"] , label="High" , color = "skyblue")
plt.title("Bar : FareGroup")
plt.xlabel("FareGroup")
plt.ylabel("Fare")
plt.grid(True,linestyle="--")
plt.legend()
```
