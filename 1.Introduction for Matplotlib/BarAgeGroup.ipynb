```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("train.csv")
df["AgeGroup"] = df["Age"].apply(lambda x : "Unknown" if pd.isnull(x)
                                 else "Child" if x < 18
                                 else "Teen" if x < 25
                                 else "Adult" if x < 50
                                 else "Senior")
agr_count = df["AgeGroup"].value_counts().reset_index()
agr_count.columns=["AgeGroup","Count"]
plt.bar(agr_count["AgeGroup"], agr_count["Count"] , color=["yellow","brown","skyblue","lightgreen","pink"] , label=["Adult","Unknown","Teen","Child","Senior"])
plt.legend(title="AgeGroups" , loc="best" , fontsize=10)
plt.title("Bar : Number of Age Groups")
plt.xlabel("Age Groups")
plt.ylabel("Number")
plt.grid(True,linestyle=":")
```
