```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np


df=pd.read_csv("train.csv")
cereal_df=pd.read_csv("cereal.csv")
cali_df = pd.read_csv("california_cities.csv")
plt.style.use('classic')
### if u can has n rows or cols , u can replace 2 to n
fig , ax = plt.subplots(nrows=2,ncols=2,figsize=(20,15))

longd = cali_df["longd"]
latd = cali_df["latd"]
size = cali_df["area_total_km2"]
color = cali_df["population_total"]

##
ax[0,0].scatter(longd,latd,c=np.log10(color),s=size,alpha=0.7,cmap="viridis")
ax[0,0].set_title("Scatter : Area & Population",color="green")
ax[0,0].set_xlabel("Longitude")
ax[0,0].set_ylabel("Latitude")
fig.colorbar(ax[0,0].collections[0],ax=ax[0,0],label="Log$_{10}$Population")
x = [50,100,250,400]
for i in x:
    ax[0,0].scatter([],[],s=i,label=str(i)+"km$^2$",color="gray",alpha=0.7)
ax[0,0].legend(title="Area",scatterpoints=1 )
ax[0,0].grid(True,linestyle="--")

###
x = np.linspace(0,10,250)
y = np.sin(x)
ax[0,1].plot(x,y,color="red" , label="Sin(x)")
ax[0,1].plot(x,np.cos(np.linspace(0,10,250)),color="yellow",label="Cos(x)")
ax[0,1].legend()
ax[0,1].axis("equal")
ax[0,1].set_xlabel("X",color="black",fontsize=10)
ax[0,1].set_ylabel("Y" , color="black", fontsize=10)
ax[0,1].set_title("Cos(x) & Sin(x)" , color="black", fontsize=15)
ax[0,1].grid(True,linestyle="--")

##
score_student = np.random.randint(0,100,300)
np.random.RandomState(0)
ax[1,0].hist(score_student,bins=30,color="lightblue")
ax[1,0].set(title="Histogram of Student Score",
            xlabel="Score",
            ylabel="Number")
ax[1,0].grid(True,linestyle="--")

##

ax[1,1].scatter(df["Age"],df["Fare"],c=df["Survived"],alpha=0.9,cmap="coolwarm") #Although Survived has only 1 or 0 :))
ax[1,1].set(title="Scatter : Age & Fare",
            xlabel="Age",
            ylabel="Fare")
fig.colorbar(ax[1,1].collections[0],ax=ax[1,1],label="Survived")
plt.xlim([0,80])
plt.ylim([0,df["Fare"].max()])
ax[1,1].grid(True,linestyle="-")
```
