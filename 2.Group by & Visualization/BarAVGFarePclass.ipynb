```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("train.csv")
bar = df.groupby("Pclass")["Fare"].mean()
plt.bar(bar.index, bar.values,color=["pink","brown","blue"],label=["P1","P2","P3"])
plt.legend()
plt.xlabel("Pclass")
plt.ylabel("AVG Fare of Pclass")
plt.title("Bar : AVG Fare of Pclass")
plt.grid(True,linestyle="--")
```
