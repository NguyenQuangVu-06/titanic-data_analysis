```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("train.csv")
df["AgeGroup"] = df["Age"].apply(lambda x : "Unknown" if pd.isnull(x)
                                 else "Child" if x < 18
                                 else "Teen" if x < 25
                                 else "Adult" if x < 50
                                 else "Senior")
bar = df.groupby("AgeGroup")["Fare"].mean()
bar.plot(kind="bar" , color=["yellow","purple","green","skyblue","lightgreen"],edgecolor="black")
plt.xlabel("Age Group")
plt.ylabel("Average Fare Age")
plt.title("AVG Fare Age Group")
plt.grid(True,linestyle="--")
```
