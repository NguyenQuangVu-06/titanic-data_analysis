```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("train.csv")
df["Embarked"]=df["Embarked"].fillna(df["Embarked"].mode()[0])
survival = df.groupby("Embarked")["Survived"].mean()
plt.bar(survival.index, survival.values,color=["pink","brown","green"],label=["C","Q","S"])
plt.legend()
plt.xlabel("Embarked")
plt.ylabel("Rate of Survival")
plt.title("Bar : Survival Rate of Embarked")
plt.grid(True,linestyle="--")
```
