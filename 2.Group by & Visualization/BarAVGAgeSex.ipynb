```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("train.csv")
bar = df.groupby("Sex")["Age"].mean()
plt.bar(bar.index, bar.values,color=["pink","yellow"],label=["female","male"])
plt.legend()
plt.xlabel("Gender")
plt.ylabel("AVG Age ")
plt.title("Bar : AVG Age of Sex")
plt.grid(True,linestyle="--")
```
