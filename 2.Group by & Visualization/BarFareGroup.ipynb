```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("train.csv")
df["FareGroup"] =df["Fare"].apply(lambda x : "Cheap" if x < 100
                                  else "Medium" if x < 200
                                  else "High")
bar = df["FareGroup"].value_counts()
plt.bar(bar.index,bar.values,color=["skyblue","lightblue","lightgreen"],label=["Cheap","Medium","High"])
plt.legend(title="Fare Group",loc="best")
plt.xlabel("FareGroup")
plt.ylabel("Number of FareGroup")
plt.title("Bar : Number of FareGroup")
plt.grid(True,linestyle="--")
```
