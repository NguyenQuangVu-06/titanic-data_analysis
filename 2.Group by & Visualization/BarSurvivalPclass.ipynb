```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("train.csv")
survival = df.groupby("Pclass")["Survived"].mean()
plt.bar(survival.index, survival.values,color=["pink","brown","blue"],label=["P1","P2","P3"])
plt.legend()
plt.xticks([1,2,3],["P1","P2","P3"])
plt.xlabel("Pclass")
plt.ylabel("Rate of Survival")
plt.title("Bar : Survival Rate of Pclass")
plt.grid(True,linestyle="--")
```
