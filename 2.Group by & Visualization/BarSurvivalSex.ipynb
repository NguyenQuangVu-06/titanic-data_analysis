```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("train.csv")
survival = df.groupby("Sex")["Survived"].mean()
plt.bar(survival.index, survival.values,color=["pink","brown"],label=["female","male"])
plt.legend()
plt.xlabel("Gender")
plt.ylabel("Rate of Survival")
plt.title("Bar : Survival Rate of Sex")
plt.grid(True,linestyle="--")
```
