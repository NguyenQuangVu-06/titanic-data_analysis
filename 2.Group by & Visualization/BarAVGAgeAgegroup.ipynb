```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("train.csv")
df["AgeGroup"] = df["Age"].apply(lambda x : "Unknown" if pd.isnull(x)
                                 else "Child" if x < 18
                                 else "Teen" if x < 25
                                 else "Adult" if x < 50
                                 else "Senior")
bar = df.groupby("AgeGroup")["Age"].mean().sort_values()
colors = {"Unknown":"yellow","Child":"purple","Teen":"green","Adult":"lightblue","Senior":"skyblue"}
for  a , v in bar.items():
    plt.bar(a,v,label=a,color=colors[a])

plt.xlabel("Age Group")
plt.ylabel("Average Age")
plt.title("Bar: AVG Age Age Group")
plt.legend()
plt.grid(True,linestyle="--")

#Not Unknown because Age of Unknown is null
```
