```python
import pandas as pd
import matplotlib.pyplot as plt
plt.style.use('classic')
df = pd.read_csv("train.csv")
line = df.sort_values(by=['Age'], ascending=False)[["Name","Age"]]
line = line.iloc[0:5]
plt.barh(line["Name"],line["Age"])
plt.xlabel("Name")
plt.ylabel("Age")
plt.title("Top 5 oldest people")
```
