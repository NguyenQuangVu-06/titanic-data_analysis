```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("train.csv")
plt.style.use('ggplot')
plt.scatter(df['Age'],df['Fare'],c=df['Survived'],cmap="viridis",alpha=0.5,edgecolors='k')
plt.colorbar(label="Survived")
plt.title("Scatter : Age & Fare " , color="yellow")
plt.xlabel("Age")
plt.ylabel("Fare")
```
