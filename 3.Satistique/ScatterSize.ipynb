```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("train.csv")
plt.style.use("classic")
a = df["Fare"]
plt.scatter(df["Age"],df["Fare"],s=a, alpha=0.7 ,edgecolor='k')
x = [0,100,200,500]
for i in x:
    plt.scatter([],[],s=i,label=str(i),alpha=0.5,color="gray")
plt.legend(scatterpoints=1,linewidth=1)
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Scatter : Age & Fare")
```
